## Exercise 3 Compute area of a circle 
Write a python script computing the area of a circle. 

The radius of the circle is provided by the user.
To provide the radius simply use the function: 
```
    radius = input(message)
```

Where message represents the name displayed before the input cell.

````python
from math import pi

radius = input("insert the radius of the circle")
# WARNING -> the input() method returns a string!!
# -> you have to cast the value

area = pi * float(radius)**2
print(f"The area of the circle with radius {radius} is {area}")

# specifying also the rpecision
print (f"The area of the circle is: {(pi * float(radius)**2):.3}")
````

## Exercise 4 Swap Java with Python 
Write a python script that and swap the word "python" with "java" and viceversa in the given string:
 
`"java is much more fun than python"
`

````python
s = "java is much more fun than python"
# CAREFUL because you cannot do the two operations in sequence (you'll end up substituting the string you just substituted)


s_new = s.replace("java", "pyth_on").replace("python", "java").replace("_", "")
# recall strings are immutable
print(s, " -> unmodified")
print(s_new)
````

## Exercise 5 Print file extension
Write a python script that output name and the extension of a file provided as input.

Note: the file doesn't have to *actually* exists

````python
filename = input("Provide a filename you like")

ext = filename.split('.')[-1]
basename = '.'.join(filename.split('.')[:-1]) # when you use split '.' are removed, you have to put them back!
print(f"The basename is {basename}, while the extension is {ext}")
````
