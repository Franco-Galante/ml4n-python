## Exercise 1 Modify a given tuple 
Write a script that modify a given tuple by writing the number 3 as the second element of a tuple
e.g. given the tuple `(10, 23, 52)` the script should return `(10, 3, 52)`


```python
t = (10, 23, 52)
t[1] = 3
```

```python
t = (10, 23, 52)
# tuple is *immutable*

t_new = t[:1] + (3,) + t[2:]  # you have to concatenate *homogeneous* elements (i.e. all have to be tuples)
print(t_new)
```

```python
tt = (e+1 for e in t)
print(tt)
```

```python
next(tt) # NO LIST COMPREHENSION EXISTS FOR TUPLES (a bit obvious) but "tt" is basically an iterator over a lazy sequence, that has been lazily generated
```

## Exercise 2 Revert the order of the elements
Return a given tuple with elements in the reverse order
e.g. given the tuple `(2, 4, 5, 8)` the script should return `(8, 5, 4, 2)`


```python
t = (2, 4, 5, 8)
r_reversed = t[::-1]

print(r_reversed)
```

## Exercise 3 Remove list duplicates 
Remove all duplicates from a given list
e.g. given the list `[1, 5, 5, 4, 2, 8, 3, 3]` the script should return `[1, 5, 4, 2, 8, 3]`

```python
l = [1, 5, 5, 4, 2, 8, 3, 3]

# for loop method
l_no_dup = []
for e in l:
    if e not in l_no_dup:
        l_no_dup.append(e)

print("List with no duplicates: ", l_no_dup)

# i prefer the version with sets
l_no_dup2 = list(set(l))
print("Other way of doing the same: ", l_no_dup2, " -> The order changed!!")
```

## Exercise 4 Euclidean distance between two vectors
Compute the euclidean distances between two vectors (passed as lists) as following:

$ dist(v1, v2) = \sqrt{\sum_i{(v1_i-v2_i)^2}} $

*hint*: you can use the `math.sqrt` function to compute the square root of a number

```python
import math

def euclidean_distance(v1, v2):
    """
    Compute the Euclidean distance between v1 and v2.
    d = sqrt( sum_i (v1_i - v2_i)^2 )
    """
    d = 0
    for v1_i, v2_i in zip(v1, v2):
        d += (v1_i - v2_i)**2

    return math.sqrt(d)

# example usage
v1 = [0, 1]
v2 = [1, 0]
print("Euclidean distance between vectors: ", euclidean_distance(v1, v2))
```
